```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_id = pd.read_parquet("/Users/deveshshah/Documents/Programming/Out_Of_Distribution_Exploration/trained_models/e5_energy_based_inference/predictions_best_val_balanced_accuracy/plantpathology_predictions.parquet")
```

```python
df_ood_flowers = pd.read_parquet("/Users/deveshshah/Documents/Programming/Out_Of_Distribution_Exploration/trained_models/e5_energy_based_inference/predictions_best_val_balanced_accuracy/flowers102_predictions.parquet")
df_ood_dtd = pd.read_parquet("/Users/deveshshah/Documents/Programming/Out_Of_Distribution_Exploration/trained_models/e5_energy_based_inference/predictions_best_val_balanced_accuracy/dtd_predictions.parquet")
df_ood_standfordcars = pd.read_parquet("/Users/deveshshah/Documents/Programming/Out_Of_Distribution_Exploration/trained_models/e5_energy_based_inference/predictions_best_val_balanced_accuracy/stanfordcars_predictions.parquet")
```

```python
id_scores = df_id["energy_score"].values
ood_flowers_scores = df_ood_flowers["energy_score"].values
ood_dtd_scores = df_ood_dtd["energy_score"].values
ood_stanfordcars_scores = df_ood_standfordcars["energy_score"].values
```

```python
plt.hist(id_scores, bins=50, alpha=0.5, label="ID (Plant Pathology)")
plt.hist(ood_flowers_scores, bins=50, alpha=0.5, label="OOD (Flowers102)")
plt.hist(ood_dtd_scores, bins=50, alpha=0.5, label="OOD (DTD)")
plt.hist(ood_stanfordcars_scores, bins=50, alpha=0.5, label="OOD (Stanford Cars)")
plt.xlabel("Energy Score")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# get mean of each distribution
print("Mean energy score for ID (Plant Pathology):", np.mean(id_scores))
print("Mean energy score for OOD (Flowers102):", np.mean(ood_flowers_scores))
print("Mean energy score for OOD (DTD):", np.mean(ood_dtd_scores))
print("Mean energy score for OOD (Stanford Cars):", np.mean(ood_stanfordcars_scores))
```
